```python
import pandas as pd

df_model2 = pd.read_csv('df_limpio_1y2.csv')
```

```python
columnas_a_eliminar = [
    'incident_state_Closed',
    'contact_type_Self service',
    'category_Category 4',
    'subcategory_Subcategory 192',
    'subcategory_Subcategory 282',
    'impact_3 - Low',
    'urgency_3 - Low',
    'priority_4 - Low'  
]


df_model2.drop(columnas_a_eliminar, axis=1)
```

Se eliminaron ciertas variables que arrojaron una correlación alta.

```python
from sklearn.metrics import accuracy_score, precision_score, log_loss, recall_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

y2 = df_model2["made_sla"]
x2 = df_model2.drop("made_sla", axis=1)
X_train, X_test, y_train, y_test = train_test_split(x2, y2, test_size=0.3, random_state=42)
model2 = LogisticRegression(max_iter=1000)
model2.fit(X_train, y_train)
y_pred2 = model2.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred2))
print("Precision:", precision_score(y_test, y_pred2))
print("Log-loss (función de pérdida):", log_loss(y_test, y_pred2))
print("Recall:", recall_score(y_test, y_pred2))

```

Como las métricas nos dan exactamente lo mismo (teniendo en cuenta la misma cantidad de decimales), concluimos que no influye en nada quitar o dejar las variables encontradas con autocorrelación, por lo que tampoco se graficará la función de pérdida, pues sería casi idéntica a la del modelo #1. Nótese que también se cambiaron las variables y las métricas tampoco cambian.
